```python
def f(x):
  return x**3+4*x**2-10   #qui podemos cambiar la funcion a la que queramos

a = 1                     #aqui podemos cambiar o variar el valor de a
b = 2                     #aqui podemos cambiar o variar el valor de b
TOL = 10**(-4)            #aqui podemos cambiar o variar el valor de TOL
```

Aqui lo que hicimos fue definir la funcion que queremos ocupar, en este caso es la del ejemplo $x^3+4x^2-10$

aparte de eso lo que hacemos es definir a las variables "a", "b" y "TOL" donde TOL es la tolerancia que se nos da en el ejercicio o la que queramos

```python

print("f(a) =", f(a)) #esto no se cambia nada mas es para evaluar, esto es
print("f(b) =", f(b)) #automatico
```

Aqui lo que hacemos solamente es evaluar la funcion en el punto "a" y "b"

```python
if f(a)*f(b) < 0: #tampoco se toca nada de aqui esto es automatico
  print("Existe un cambio de signo.")
else:
  print("No existe un cambio de signo, por tanto no cumple el TVI.")
```

Lo que estamos haciendo aqui es verificar si existe un cambio de signo, para verificar que si se cumple el $TVI$

```python
p = a + ( b - a ) * 0.5

print("p = ",p)
```

Aqui lo que estamos haciendo es calcular el valor medio del intervalo aplicando el algoritmo de burden para despues hacier el cambio $[a,b] -> [a,p]$

```python
FP = f(p)
FA = f(a)
FB = f(b)

print("f(p) = ",FP)

if FP == 0 or (b - a) * 0.5 < TOL:
    print("p es una aproximacion de la raiz:", p)

```

Lo que estamos haciendo practicamente es solo verificar si p es raiz o no

```python
if FA * FP > 0:
    a = p
    FA = FP
else:
    b = p
```

Lo que estamos haciendo aqui es decir que si $f(a)$ y $f(p)$ tienen el mismo signo entonces se cambia $b -> p$ en caso contrario seria $p -> a$

```python
i = 1
N0 = 100
FA = f(a)

iteraciones = []
valores_a = []
valores_b = []
valores_p = []
valores_fp = []

while i <= N0:

    p = a + (b - a) / 2
    FP = f(p)

    iteraciones.append(i)
    valores_a.append(a)
    valores_b.append(b)
    valores_p.append(p)
    valores_fp.append(FP)

    if FP == 0 or (b - a) * 0.5 < TOL:
       print("Aproximación de la raíz:", p)
       break

    i = i + 1

    if FA * FP > 0:
        a = p
        FA = FP
    else:
        b = p




```

en este bloque del codigo lo que hicimos fue practicamente hacer la base del codigo, aqui el proceso que hicimos al inicio se repite consecutivamente hasta que $f(p)=0$ ó  $(b-a) * 0.5 < TOL$

```python
import pandas as pd

tabla = pd.DataFrame({
    "n": iteraciones,
    "a": valores_a,
    "b": valores_b,
    "p": valores_p,
    "f(p)": valores_fp
})

tabla.style \
    .format({
        "a": "{:.6f}",
        "b": "{:.6f}",
        "p": "{:.6f}",
        "f(p)": "{:.6f}"
    }) \
    .set_caption("Iteraciones del método de bisección") \
    .set_properties(**{
        "text-align": "center"
    })
```
