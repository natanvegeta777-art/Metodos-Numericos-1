```python
def f(x):
  return x**3 + 4*x**2 - 10 #basicamente definimos la funcion original
```

```python
#el intervalo [a,b]=[1,2] portanto definimos "a" y "b"
a = 1
b = 2

#despues verificamos si en este intervalo si hay raiz por TIV
#definimos f(a) y f(b) si hay cambio de signo si hay raiz y podemos seguir

f(a)
f(b)

print("f(a) = ", f(a))
print("f(b) = ", f(b))
```

```python
if f(a)*f(b) < 0:  #verificamos la condicion para que pueda avanzar el programa
                   #si hay cambio de signo si avanza si no pues no se cumple TVI
  print("Existe un cambio de signo.")
else:
  print("No existe un cambio de signo, por tanto no cumple el TVI.")
```
