En este programa implementaremos el metodo de punto fijo ocupando el ejemplo numero 1 de la pagina 35 del libro del burden

```python
def f(x):
  return x**3 + 4*x**2 - 10 #basicamente definimos la funcion original
```

```python
#el intervalo [a,b]=[1,2] portanto definimos "a" y "b"
a = 1
b = 2

#despues verificamos si en este intervalo si hay raiz por TIV
#definimos f(a) y f(b) si hay cambio de signo si hay raiz y podemos seguir

f(a)
f(b)

print("f(a) = ", f(a))
print("f(b) = ", f(b))
```

```python
if f(a)*f(b) < 0:  #verificamos la condicion para que pueda avanzar el programa
                   #si hay cambio de signo si avanza si no pues no se cumple TVI
  print("Existe un cambio de signo.")
else:
  print("No existe un cambio de signo, por tanto no cumple el TVI.")
```

```python
#definimos la funcino g(x) para construie la sucesion x_n
#esta funcion se deduce al despegar a x de la forma x=g(x)


def g(x):
    return (10 / (x + 4))**0.5
```

```python
from os import TFD_NONBLOCK
#definimos 3 variables basicas para poder trabajar

X0  = 1 #VALOR INICIAL
TOL = 0.0001 #TOLERANCIA
n0  = 100 #NUMERO MAXIMOM DE ITERACIONES
```

```python



i = 1
iteraciones = []

while i <= N0: #las iteraciones no pueden se mas de 100
    p = g(X0)  #evaluamos la funcion en el punto incicial
    error = abs(p - X0) #aqui vemos el valor eroor

    iteraciones.append([i, X0, p, f(p), error]) #para crear la tabla

    if error < TOL: #verificar cuando parar
        break

    X0 = p #hacemos que p cambie de valor constante mente para que converga
    i = i + 1 #aumentar las iteraciones pa que no sea infinto
print("Aproximación de la raíz:", p)
print("Error final:", error)
print("Iteraciones realizadas:", i)
print("Evaluación de f(p):", f(p))
```

```python

print("n         Xn          Xn+1        f(p)        Error")  #damos un tipo de ormato de tabla

for fila in iteraciones:
    print(fila)
```
